```python
# import libraries

import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense, Bidirectional, LSTM, Input
```

```python
# load IMDb Data
vocab_size = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

# Padding sequences

max_length = 100

X_train = pad_sequences(X_train, maxlen=max_length)
X_test = pad_sequences(X_test, maxlen=max_length)
```

### **Simple RNN**

```python
rnn_model = Sequential()

# Input layer
rnn_model.add(Input(shape=(max_length,)))

# Embedding layer
rnn_model.add(Embedding(input_dim=vocab_size, output_dim=64))

# Simple RNN
rnn_model.add(SimpleRNN(32))

# Output layer
rnn_model.add(Dense(1, activation='sigmoid'))

# Compile model
rnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# model Summary
rnn_model.summary()

# train model
history_rnn = rnn_model.fit(X_train, y_train, epochs = 2 , batch_size = 64, validation_split=0.2)

# Evaluate model
loss_rnn, acc_rnn = rnn_model.evaluate(X_test, y_test)
print("\nSimple RNN Test Accuracy:", acc_rnn)
```

```python
birnn_model = Sequential()

# Input layer
birnn_model.add(Input(shape=(max_length,)))

# Embedding layer
birnn_model.add(Embedding(input_dim=vocab_size, output_dim=64))

# Bidirectional RNN
birnn_model.add(Bidirectional(SimpleRNN(32)))

# Output layer
birnn_model.add(Dense(1, activation='sigmoid'))

# Compile model
birnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# model Summary
birnn_model.summary()
```

```python
# Train model
history_birnn = birnn_model.fit(X_train, y_train, epochs=2, batch_size=64, validation_split=0.2)
```

```python
# Evaluate model
loss_birnn, acc_birnn = birnn_model.evaluate(X_test, y_test)

print("\nBidirectional RNN Test Accuracy:", acc_birnn)
```
