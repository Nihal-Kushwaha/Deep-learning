# **Word2Vec: CBOW vs Skip-Gram**

In this , we used a text dataset to train Word2Vec models using two approaches: **CBOW** and **Skip-Gram**. The text was first preprocessed using tokenization and lowercasing, and then both models were trained to learn word relationships in the form of vectors.

```python
import re
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
import seaborn as sns
import matplotlib.pyplot as plt
import gensim
from gensim.models import Word2Vec
from sklearn.decomposition import PCA
```

## **Loading & Preprocessing the Text Dataset**

```python
txt = [
  "Data scientists use statistical methods to derive meaningful insights daily",
  "GPU acceleration significantly speeds up the training of neural networks",
  "Intelligent agents can perform automated tasks with very high precision",
  "Distributed systems like PySpark handle large scale data processing efficiently",
  "Backpropagation updates neural network weights by calculating the gradient descent",
  "Feature engineering involves transforming raw variables into more predictive inputs",
  "Exploratory data analysis reveals underlying distributions through visual plotting techniques",
  "Hyperparameter tuning optimizes model performance by adjusting learning rate values",
  "Topic modeling identifies latent themes across a collection of documents",
  "Random forests use an ensemble of decision trees to reduce overfitting"
]
```

```python
processed_data = []

for sentence in txt:
    tokens = word_tokenize(sentence.lower())  # lowercase + tokenize
    processed_data.append(tokens)

print("Sample Preprocessed Data:", processed_data[0])
```

## **Train CBOW Model (sg=0)**

```python
# CBOW Model
cbow_model = Word2Vec(
    sentences=processed_data,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0
)
```

## **Train Skip-Gram Model (sg=1)**

```python
# Skip-Gram Model
skipgram_model = Word2Vec(
    sentences=processed_data,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1
)
```

## **Vocabulary Size**

```python
print("\nCBOW Vocabulary Size:", len(cbow_model.wv))
print("Skip-Gram Vocabulary Size:", len(skipgram_model.wv))
```

The vocabulary size for both CBOW and Skip-Gram models is the same because both models are trained on the same preprocessed dataset with identical parameters (min_count=1). The training algorithm (CBOW or Skip-Gram) affects how word vectors are learned, but not which words are included in the vocabulary.

## **Word Vectors (5 words)**

```python
words = ["data", "neural", "model", "learning", "analysis"]

print("\n CBOW Word Vectors")
for word in words:
    if word in cbow_model.wv:
        print(f"{word}: {cbow_model.wv[word][:5]}")  # first 5 values

print("\n Skip-Gram Word Vectors")
for word in words:
    if word in skipgram_model.wv:
        print(f"{word}: {skipgram_model.wv[word][:5]}")
```

## **Similar Words**

```python
query_word = "data"

print("\nTop 5 similar words (CBOW):")
print(cbow_model.wv.most_similar(query_word, topn=5))

print("\nTop 5 similar words (Skip-Gram):")
print(skipgram_model.wv.most_similar(query_word, topn=5))
```

### **Both CBOW and Skip-Gram models produced the same vocabulary size because they were trained on the same dataset. However, their word vectors and similar word results differ due to different learning methods. CBOW is faster and works well for small data, while Skip-Gram captures better relationships, especially for less frequent words.**
